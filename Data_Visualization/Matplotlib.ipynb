```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Matplotlib Basic

```python
x=np.linspace(0,5,11)
x
y=x**2
y
```

# Line Plot

A line plot connects data points to show trends over continuous variables like time or sequence. It is ideal for tracking changes such as stock prices or temperature variations.

```python
plt.plot(x,y)
```

```python
plt.plot(x,y,label="exponential")
plt.title("demo")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.legend()
plt.show()
```

# Subplot in Matplotlib


subplot devide the canvas into parts


```python
plt.subplot(2,2,1)# will devide the canvas into (Row,columns,subplot you want to select)
# to plot a chart in subplot do plotting just under the subplot line 
plt.subplot(2,2,2) 
plt.plot(x,y)
plt.subplot(2,2,3) 
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y**2)
```

# object oriented way

it is used to acces the canvas and modify it [full control over canvas unlike subplot which have fixed canvas ]

```python
fig=plt.figure(figsize=(20,8),dpi=100) # figsize(length , height) it will decide main canvas size  
axis1=fig.add_axes([0.1,0.1,.4,.8]) #[move from left,move from bottom,length, height] it will decide sub canvas
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.3,0.3,0.3]) # multiple plot can be created on same canvas and modifyed using same object here fig
axis2.plot(x,y)
```

# Types of  plot

### Bar Chart

Good for comparing categorical data (sales across products, student marks in subjects).


```python
categories = ['A', 'B', 'C', 'D']
values = [10, 24, 36, 18]
plt.bar(categories, values, color='green') # for horizontal use barh
plt.title("Bar Chart")
plt.show()
```

## Histogram


Used to show distribution of data (frequency of values). 

Bins in a histogram are adjacent, non-overlapping intervals that divide a range of data values to count how many data points fall into each segment

```python
data = np.random.randn(1000)
plt.hist(data, bins=30,color='purple',edgecolor='black')
plt.title("Histogram")
plt.show()
```

## Scatter Plot


Used to find relationships/correlation/varience between two variables. Example: height vs weight. 

```python
v = np.random.rand(50)
p = np.random.rand(50) 
plt.scatter(x, y, color='red', marker='x')
plt.title("Scatter Plot")
plt.show()
```

## Pie Chart

Good for showing percentage/proportion of categories in a whole.

```python
sizes = [20, 30, 25, 25] 
labels = ['A', 'B', 'C', 'D'] 
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart") 
plt.show()
```

## Box Plot

Useful for detecting outliers and spread of data. 

```python
data = [7, 8, 5, 6, 7, 7, 8, 5, 6, 9, 10] 
c=plt.boxplot(data) 
plt.title("Box Plot") 
plt.show(c)

```

## Heatmap

Good for showing correlation matrix between multiple variables.

 It use colours to represent correlation matrices between variables. Useful in data science to quickly spot strong positive or negative relationships.
