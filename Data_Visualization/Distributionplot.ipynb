##### Some plot which help in visualization of quantative data set

* distplot/histplot

* joinplot

* pairplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('tips') # seabon have some in built data set
df
```

```python
df["size"].unique()
```

# Distplot/histplot


Bins devide the range into adjcent, non-overlaping, equal parts

```python
sns.displot(df['total_bill'],bins=20,kde=True) # by defalt bin=20
# Kde is kernal density estimater [you can know more about it in statstic part]
# curve line is kde
```

```python
#to make multiple plot use subplot
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins=20,kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=20,kde=True)
```

# Joinplot


```python
sns.jointplot(x='total_bill',y='tip',data=df,kind="scatter")
```

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind="kde")
```

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind="hist")
```

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind="reg")  #lines on scatter plot is reggeration line
#curve on hiostogram is also showing reggeration
```

# Pairplot

A pair plot (also known as a scatterplot matrix) is a powerful data visualization tool used in statistics and data science to visualize relationships between multiple numerical variables in a dataset.

```python
sns.pairplot(df,hue='sex')  #didive the data category wise          
```

```python
sns.pairplot(df,palette='rainbow') # Ues palette with hue other wise you will get error
```

```python
sns.pairplot(df,hue='size',palette='rainbow')
```
